# 01 - Bronze Layer

Loads the three CSV files and stores them as raw Parquet-backed logical tables.

Bronze keeps the source values as strings and adds ingestion metadata.

```python
from setup import get_spark_session
from config import (
    SALES_ORDER_DETAIL_PATH,
    SALES_ORDER_HEADER_PATH,
    PRODUCTS_PATH,
    BRONZE_LAYER,
    BRONZE_TABLES,
)
from utils import (
    read_csv_raw,
    save_table,
    add_ingestion_metadata,
    validate_input_file,
    show_title,
    print_table_location,
)

spark = get_spark_session()
```

```python
show_title("Bronze Layer - Validating input files")

for path in [SALES_ORDER_DETAIL_PATH, SALES_ORDER_HEADER_PATH, PRODUCTS_PATH]:
    validate_input_file(path)
    print(f"Input file found: {path}")
```

```python
show_title("Bronze Layer - Reading CSV files")

bronze_sales_order_detail = add_ingestion_metadata(
    read_csv_raw(spark, SALES_ORDER_DETAIL_PATH),
    SALES_ORDER_DETAIL_PATH,
)

bronze_sales_order_header = add_ingestion_metadata(
    read_csv_raw(spark, SALES_ORDER_HEADER_PATH),
    SALES_ORDER_HEADER_PATH,
)

bronze_products = add_ingestion_metadata(
    read_csv_raw(spark, PRODUCTS_PATH),
    PRODUCTS_PATH,
)
```

```python
show_title("Bronze Layer - Saving raw Parquet tables")

save_table(bronze_sales_order_detail, BRONZE_LAYER, BRONZE_TABLES["sales_order_detail"])
save_table(bronze_sales_order_header, BRONZE_LAYER, BRONZE_TABLES["sales_order_header"])
save_table(bronze_products, BRONZE_LAYER, BRONZE_TABLES["products"])

for table_name in BRONZE_TABLES.values():
    print_table_location(BRONZE_LAYER, table_name)
```

```python
show_title("Bronze Layer - Sample")

bronze_sales_order_detail.show(5, truncate=False)
bronze_sales_order_header.show(5, truncate=False)
bronze_products.show(5, truncate=False)
```
